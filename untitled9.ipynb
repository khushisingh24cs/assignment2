```python
from sklearn.neighbors import KNeighborsClassifier
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
import numpy as np
import pandas as pd
from sklearn.impute import SimpleImputer
from sklearn.metrics import  mean_squared_error ,r2_score
from sklearn.linear_model import LinearRegression
import matplotlib.pyplot as plt

```

```python
df=pd.read_excel(r"C:\Users\khush\OneDrive\Desktop\Housing.xlsx") # r used to read / as an eascape char not new line
```

```python
df.head()     #data frame 2d representation
```

```python
# select feature and target
X=df[['area']] #independent
Y=df['price']    # dependent       # ravel 2d to 1d
```

```python
# test train split data set
X_train, X_test, y_train, y_test = train_test_split(
    X,
    Y, 
    test_size=0.2,     #20% test
    random_state=42  #ensure split maintain 
)
```

```python
# create linear regression model
model=LinearRegression()
```

```python
# train the model
model.fit(X_train,y_train)
```

```python
# make predict
y_pred = model.predict(X_test)

```

```python
result = pd.DataFrame({
    "actual price": y_test.values,
    "predicted price": y_pred          #actual vs predicted value
})
```

```python
print(result)
```

```python
# print equation value
print("\nSlope (coff):",model.coef_[0])
print("intercept:",model.intercept_)
```

```python
# evalute model
mse=mean_squared_error(y_test,y_pred)
r2=r2_score(y_test,y_pred)
```

```python
print("\n mean squared error:",mse)
print("r2 error:",r2)     #poor model :
```

```python
plt.figure(figsize=(8,6))
plt.scatter(X_train,y_train,
color="blue",label="actual data"
)
plt.plot(X_train,model.predict(X_train),color="red",linewidth=2,label="predicted data")
plt.title("area vs price")
plt.xlabel("area")
plt.ylabel("price")           
plt.legend()
plt.show()
```
