```python
missing_percentage=(df.isnull().sum())/len(df)*100
print(missing_percentage)
```

```python
age_median=df["Life expectancy"].median
df["Life expectancy"]=df["Life expectancy"].fillna(age_median)
```

```python
# Fill missing values in a categorical column
status_mode = df["Status"].mode()[0]
df["Status"] = df["Status"].fillna(status_mode)
```

```python
df=df.drop(columns=["Country"])
```

```python
df.isnull().sum()
```

```python

duplicate_row=df.duplicated().sum()
print(duplicate_row)
```

```python
Q1 = df["GDP"].quantile(0.25)
Q3 = df["GDP"].quantile(0.75)

IQR = Q3 - Q1

lb = Q1 - 1.5 * IQR
ub = Q3 + 1.5 * IQR

df = df[(df["GDP"] >= lb) & (df["GDP"] <= ub)]
```
